# Task 4 – Cross-Validation & Hyperparameter Tuning

```python
from sklearn.model_selection import StratifiedKFold,GridSearchCV,RandomizedSearchCV
```

```python
cv=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc=RandomForestClassifier()
rf_cross={
    'n_estimators':[50,100,50],
    'max_depth':[3, 4, 5],
    'min_samples_split':[2, 4]
}
```

```python
rf_val=RandomizedSearchCV(rfc,rf_cross,cv=cv,scoring='accuracy',verbose=1,n_jobs=-1)
rf_val.fit(X_train_scaled,Y_train)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier()
knn_cross={
    'n_neighbors':[3,5]
}
```

```python
knn_val=RandomizedSearchCV(knn,knn_cross,cv=cv,scoring='accuracy',verbose=1,n_jobs=-1)
knn_val.fit(X_train_scaled,Y_train)
```

```python
from sklearn.tree import DecisionTreeClassifier
des=DecisionTreeClassifier()
dec_cross={
    'criterion':['gini','entropy'],
    'max_depth':[3,4,5]
}
```

```python
dec_val=RandomizedSearchCV(des,dec_cross,cv=cv,scoring='accuracy',verbose=1,n_jobs=-1)
dec_val.fit(X_train_scaled,Y_train)
```

```python
from sklearn.naive_bayes import GaussianNB
NB=GaussianNB()
NB_cross={
    'var_smoothing':[1e-9,1e-10,1e-11]
}
```

```python
NB_val=GridSearchCV(NB,NB_cross,cv=cv,scoring='accuracy',verbose=1,n_jobs=-1)
NB_val.fit(X_train_scaled,Y_train)
```

```python
from sklearn.svm import LinearSVC
svm=LinearSVC(dual=False,random_state=42,max_iter=1000,tol=1e-3)
svm_cross={
    'C':[0.1,1,10]
}

```

I use the LinearSVC because normally in kernel use the method means it take so much time and still running so use this and set a iteration also

```python
svm_val=RandomizedSearchCV(svm,svm_cross,cv=cv,scoring='accuracy',verbose=1,n_jobs=-1)
svm_val.fit(X_train_scaled,Y_train)
```

# Task 5 – Model Comparison & Selection

```python
#Random Forest Classifier model with the best parameters
rf_cr=RandomForestClassifier(max_depth=5, min_samples_split=4, n_estimators=50)
rf_cr.fit(X_train_scaled,Y_train)
y_rf_pred=rf_cr.predict(X_test_scaled)
print(f"Accuracy:{accuracy_score(Y_test,y_rf_pred)}")
```

```python
#KNeighbors Classifier model with the best parameters
knn=KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train_scaled,Y_train)
y_knn_pred=knn.predict(X_test_scaled)
print(f"Accouracy:{accuracy_score(Y_test,y_knn_pred)}")
```

```python
#Decision Tree Classifier model with the best parameters
dec=DecisionTreeClassifier(criterion='entropy', max_depth=5)
dec.fit(X_train_scaled,Y_train)
y_dec_pred=dec.predict(X_test_scaled)
print(f"Accuracy:{accuracy_score(Y_test,y_dec_pred)}")
```

```python
#Naive Bayes model in GaussianNB with the best parameters
NB=GaussianNB()
NB.fit(X_train_scaled,Y_train)
y_nb_pred=NB.predict(X_test_scaled)
print(f"Accuracy:{accuracy_score(Y_test,y_nb_pred)}")
```

```python
#Support Vector Machine with the best parameters
svm=LinearSVC(C=0.1, dual=False, random_state=42, tol=0.001)
svm.fit(X_train_scaled,Y_train)
y_svm_pred=svm.predict(X_test_scaled)
print(f"Accuracy:{accuracy_score(Y_test,y_svm_pred)}")
```

```python
model_dict={
    'Random Forest Classifier':(rf_val.best_estimator_,X_test_scaled),
    'KNeighbors Classifier':(knn_val.best_estimator_,X_test_scaled),
    'Decision Tree Classifier':(dec_val.best_estimator_,X_test_scaled),
    'Naive Bayes':(NB_val.best_estimator_,X_test_scaled),
    'Support Vector Machine':(svm_val.best_estimator_,X_test_scaled)
}
model_dict
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,confusion_matrix
result=[]
for name,(model_obj,X_test_scaled) in model_dict.items():
  pred=model_obj.predict(X_test_scaled)
  result.append({
      'Model':name,
      'Accuracy':accuracy_score(Y_test,pred),
      'Precision':precision_score(Y_test,pred),
      'Recall':recall_score(Y_test,pred),
      'F1 Score':f1_score(Y_test,pred),
      'Confusion Matrix':confusion_matrix(Y_test,pred)
  })
```

Instead of predict all models use a for loop we know the range and use it and predict all models

```python
compare_df=pd.DataFrame(result).sort_values(by="Accuracy",ascending=False)
compare_df
```

```python
best_model_name=compare_df.iloc[0,0]
best_model=model_dict[best_model_name][0]
best_X_test=model_dict[best_model_name][1]
```

```python
best_pred=best_model.predict(best_X_test)
confusion_matrix=confusion_matrix(Y_test,best_pred)
```

```python
sns.heatmap(confusion_matrix,annot=True,fmt='d')
plt.title('Support Vector Machine')
plt.show()
```

```python
print("Best Model Name:",best_model_name)
print("Best Model:",best_model)
```

Therefore Support Vector Machine is the best model compare to all models and use the best parameter get from  hyper parameter tuning and cross validation
